```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("../data/mesures_capteurs.csv")
print(df)
```

### Partie 1 – Graphique linéaire (Line Plot) 

##### 1) Afficher l'évolution de la température en fonction du temps, avec le titre, les labels, la légende et la grille.

```python
plt.figure (figsize =(10, 6) )
plt.plot (df['date_heure'], df['temperature'])
plt.title ('evolution de la temperature en fonction du temps')
plt.xlabel ('temps')
plt.ylabel ('temperature')
plt.grid (True)
plt.show()
```

##### 2) Identifier visuellement les valeurs qui semblent anormales.

***explication*** : les valeurs qui semblent anormales sont les temperatures entre -19 et -20 et entre 59 et 60 qui sont des valeurs aberrantes

### Partie 2 – Diagramme en barres (Bar Chart) 

##### 1) Déterminer consommation_batiment qui stocke les consommations moyennes des bâtiments 

```python
consommation_batiment = df.groupby('batiment')['consommation'].mean()
print(consommation_batiment)
```

##### 2) Comparer graphiquement la consommation moyenne des bâtiments 

```python
plt.figure(figsize=(10, 6))
plt.bar(consommation_batiment.index, consommation_batiment.values, color='blue', edgecolor='black', width=0.6)

# 3. Titre, labels, légende et grille
plt.title("Comparaison de la consommation moyenne par bâtiment", fontsize=12 )
plt.xlabel("Bâtiments", fontsize=12)
plt.ylabel("Consommation moyenne (kWh)", fontsize=12)
 # Grille uniquement sur l'axe vertical Y
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 5. Afficher le graphique
plt.show()
```

##### 3) Comparer la consommation moyenne des bâtiments avec un graphique horizontal 

```python
plt.figure(figsize=(10, 6))

plt.barh(consommation_batiment.index, consommation_batiment.values, color='red', edgecolor='black', height=0.6)

# 3. Titre, labels et grille
plt.title("Comparaison de la consommation moyenne par bâtiment", fontsize=12)
plt.xlabel("Consommation moyenne (kWh)", fontsize=12)
plt.ylabel("Bâtiments", fontsize=12)

# La grille s'applique cette fois-ci sur l'axe horizontal (X)
plt.grid(axis='x', linestyle='--', alpha=0.7)


# 5. Afficher le graphique
plt.show()
```

##### 4) Dans quels cas un graphique horizontal peut-il être plus lisible ? 

***explication***: un graphique horizontal peut etre plus lisible lorsque qu'on a plusieurs variables comme exemple ici des les batiments.

### Partie 3 – Histogramme

##### 1) Afficher la distribution des températures en divisant la plage des données en 20 classes et en affichant uniquement les lignes horizontales de la grille 

```python
plt.figure(figsize=(10 , 6))

plt.hist(df['temperature'], bins=20, color='b', edgecolor ='black', alpha =0.7)
plt.title('distribution des temperatures' , fontsize=12)
plt.xlabel('temperature', fontsize=12)
plt.ylabel('frequence', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

##### 2)a) où se concentrent les températures 

Les temperatures se concentrent au milieu dans l'intervalle [13,36] , qu'on peut dire des temperatures normales.

##### 3 b) si la distribution semble symétrique 

```python
print("Moyenne :", df['temperature'].mean())
print("Médiane :", df['temperature'].median())
```

Apres notre analyse nous pouvons dire que la distribution est symetrique pour 3 raisons : notre histogramme est en forme cloche c'est a dire on a une barre haute qui se situe au centre du graphe. 
Et lorqu'on trace par exemple une ligne imaginaire au milieu du pic , la moitie a gauche et a droite sont pratiquement identiques . et quand on a calculer la mediane et la moyenne de la temperature pour avoir une certitude et on voit que les deux valeurs sont pratiquement identiques . donc on peut dire que la distribution est symetrique.

##### c) s'il existe des valeurs très éloignées 

Apres notre analyse , il existe des valeurs extremement eloignes (proche de -20 degre et proche de 60 degre).

##### d) si des valeurs aberrantes sont visibles. 

oui nous pouvons voir qu'il ya des valeurs aberrantes environ -15 et 56 degre

##### 3) Afficher la distribution de la consommation en divisant la plage des données en 20 classes

```python
plt.figure(figsize=(10 , 6))

plt.hist(df['consommation'], bins=20, color='g', edgecolor ='black', alpha =0.7)
plt.title('distribution de la consommation' , fontsize=12)
plt.xlabel('consommation', fontsize=12)
plt.ylabel('frequence', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
